# K-Means — Finding Clusters in Unlabeled Data

Recap from Week 2: K-Means repeatedly (1) assigns each point to its nearest
center, then (2) recomputes each center as the average of its assigned points,
until nothing changes. Unlike the earlier notebooks, this one has **no labels at
all** — that's the point, it's unsupervised. We generate a synthetic dataset made
specifically for teaching clustering, so the groups are visually obvious and you
can directly check whether K-Means found what you'd expect.

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
```

```python
# Generate 4 visually distinct blobs of points -- notice there are no labels used below
X, true_cluster = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=42)

plt.scatter(X[:, 0], X[:, 1])
plt.title("Raw data K-Means will see (no labels)")
plt.show()
```

```python
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
kmeans.fit(X)
predicted_cluster = kmeans.predict(X)

plt.scatter(X[:, 0], X[:, 1], c=predicted_cluster, cmap="viridis")
plt.scatter(
    kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
    c="red", marker="X", s=200, label="centers"
)
plt.title("K-Means result (k=4)")
plt.legend()
plt.show()
```

### What to notice

- We told K-Means `n_clusters=4` — it doesn't discover that number on its own,
  it just organizes the data into however many groups you ask for. This is
  exactly the "you have to choose k yourself" limitation from Week 2.
- The red X marks are the final cluster centers after the assign/recompute loop
  converged.

**Try this:** change `n_clusters` to `2` and then to `8` (keep `centers=4` in
`make_blobs`), and re-run. What does K-Means do when you ask for the wrong number
of clusters? There's no "right answer" it can fall back to — it will always
produce exactly the number of groups you asked for, even if that number is wrong.
